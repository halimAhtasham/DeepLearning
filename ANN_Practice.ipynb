```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 1. Generate dummy classification data (1000 samples, 20 features)
X_train = np.random.random((1000, 20))
y_train = np.random.randint(2, size=(1000, 1))

# 2. Build the ANN architecture
model = Sequential([
    # Input layer implicit; First hidden layer has 64 neurons
    Dense(64, activation='relu', input_shape=(20,)),
    # Second hidden layer
    Dense(32, activation='relu'),
    # Output layer for binary classification
    Dense(1, activation='sigmoid')
])

# 3. Compile the network
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# 4. Train the model
model.fit(X_train, y_train, epochs=10, batch_size=32)

# 5. Make predictions
predictions = model.predict(X_train[:5])
print("Sample Predictions:\n", predictions)
```
