```python
import numpy as np

# Define the sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Initialize weights and biases for a simple neural network
# Assuming 2 input features, 2 hidden units, and 1 output unit for XOR
np.random.seed(0) # for reproducibility
W1 = np.random.randn(2, 2) # Weights for hidden layer (2 inputs, 2 hidden units)
b1 = np.zeros((1, 2))      # Biases for hidden layer (1 row, 2 columns)
W2 = np.random.randn(2, 1) # Weights for output layer (2 hidden units, 1 output)
b2 = np.zeros((1, 1))      # Biases for output layer (1 row, 1 column)

# Sample input (XOR truth table - first row)
X_sample = np.array([[0, 1]])  # Input: x1=0, x2=1

# Hidden Layer calculation
z1 = np.dot(X_sample, W1) + b1
a1 = sigmoid(z1)  # Using our sigmoid function from earlier

# Output Layer calculation
z2 = np.dot(a1, W2) + b2
y_hat = sigmoid(z2)

print(f"Input: {X_sample}")
print(f"Hidden layer output (a1): {a1}")
print(f"Final Prediction (y_hat): {y_hat[0][0]:.4f}")
# At initialization (random weights), this will output ~0.5 (random guess).
# After training, it will become 1.0 for (0,1) XOR.
```
