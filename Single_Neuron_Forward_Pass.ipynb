```python
import numpy as np

# Input features (3 features)
x = np.array([0.5, 0.8, -0.2])
# Weights (synaptic strength)
w = np.array([0.4, -0.3, 0.9])
# Bias (threshold offset)
b = 0.1

# Step 1: Weighted Sum
z = np.dot(x, w) + b  # np.dot does (0.5*0.4 + 0.8*-0.3 + -0.2*0.9) + 0.1
print(f"Linear Sum (z): {z:.4f}")

# Step 2: Activation (Using Sigmoid to squash between 0 and 1)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

output = sigmoid(z)
print(f"Neuron Output (a): {output:.4f}")
```
